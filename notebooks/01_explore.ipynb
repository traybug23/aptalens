```python
def compute_mcrmse(y_true, y_pred):
    """
    y_true, y_pred: dict or list of arrays/matrices for each target column
    """
    rmses = []
    for col in y_true:
        mse = np.mean((np.array(y_true[col]) - np.array(y_pred[col])) ** 2)
        rmses.append(np.sqrt(mse))
    return np.mean(rmses)


# Toy test case
dummy_true = {"reactivity": [0.1, 0.5, 0.2], "deg_Mg_pH10": [0.0, 1.0, 0.4]}
dummy_pred = {"reactivity": [0.2, 0.4, 0.2], "deg_Mg_pH10": [0.1, 0.8, 0.5]}
print("Toy MCRMSE:", compute_mcrmse(dummy_true, dummy_pred))
```

```python
# a) Is `index` just a row number?
print((df["index"] == np.arange(len(df))).all())
df = df.drop(columns="index")

# b) Where do the extreme values live?
targets = ["reactivity", "deg_Mg_pH10", "deg_Mg_50C"]
for t in targets:
    worst = df[t].apply(lambda a: np.abs(a).max()).sort_values(ascending=False).head(5)
    print(t)
    print(df.loc[worst.index, ["id", "signal_to_noise", "SN_filter"]].assign(max_abs=worst), "\n")

# c) Share of negative and NaN values
for t in targets:
    v = np.concatenate(df[t].values)
    print(t, "negative:", round((v < 0).mean(), 4), "nan:", int(np.isnan(v).sum()))

# d) What does SN_filter actually encode?
pairs = {
    "reactivity": "reactivity_error",
    "deg_Mg_pH10": "deg_error_Mg_pH10",
    "deg_pH10": "deg_error_pH10",
    "deg_Mg_50C": "deg_error_Mg_50C",
    "deg_50C": "deg_error_50C",
}
min_all = df[list(pairs)].apply(lambda r: min(np.min(r[c]) for c in pairs), axis=1)
sn_calc = df.apply(
    lambda r: np.mean([np.mean(r[t]) / np.mean(r[e]) for t, e in pairs.items()]), axis=1
)
print(pd.crosstab(df["SN_filter"], [min_all > -0.5, sn_calc > 1.0]))
print("my S/N vs provided S/N correlation:", np.corrcoef(sn_calc, df["signal_to_noise"])[0, 1])


# e) Baseline: predict the training mean, on all rows and on filtered rows
def stack(frame):
    return np.column_stack([np.concatenate(frame[t].values) for t in targets])


for name, sub in [("all rows", df), ("SN_filter == 1", df[df["SN_filter"] == 1])]:
    Y = stack(sub)
    pred = np.tile(Y.mean(axis=0), (len(Y), 1))
    print(name, "mean-predictor MCRMSE:", np.sqrt(((Y - pred) ** 2).mean(axis=0)).mean().round(4))

# f) Fixed positions across sequences (shared design regions)
seq_arr = np.array([list(s) for s in df["sequence"]])
cons = [(seq_arr[:, i] == pd.Series(seq_arr[:, i]).mode()[0]).mean() for i in range(107)]
print("positions with >95% identical base:", [i for i, v in enumerate(cons) if v > 0.95])
```
